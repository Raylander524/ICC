### 6.  Teste de Normalidade

```python
from scipy.stats import shapiro, normaltest, kstest, zscore
```

```python
# Seleciona apenas colunas numéricas
df_numerico = df.select_dtypes(include='number')

# Aplica os testes
for col in df_numerico.columns:
    dados = df[col].dropna()

    # Shapiro-Wilk (mais confiável para amostras menores)
    if len(dados) <= 5000:
        stat, p = shapiro(dados)
        print(f"[Shapiro-Wilk] {col}: estatística={stat:.4f}, p-valor={p:.4f}")
        if p > 0.05:
            print("  → Distribuição normal (não rejeita H0)\n")
        else:
            print("  → Distribuição NÃO normal (rejeita H0)\n")

    # D’Agostino-Pearson (mais adequado para > 20 elementos)
    if len(dados) >= 20:
        stat, p = normaltest(dados)
        print(f"[D’Agostino-Pearson] {col}: estatística={stat:.4f}, p-valor={p:.4f}")
        if p > 0.05:
            print("  → Distribuição normal (não rejeita H0)\n")
        else:
            print("  → Distribuição NÃO normal (rejeita H0)\n")

    # Kolmogorov-Smirnov (com z-score normalizado)
    dados_z = zscore(dados)  # Normaliza os dados para média 0 e desvio 1
    stat_ks, p_ks = kstest(dados_z, 'norm')
    print(f"  [Kolmogorov-Smirnov] estatística={stat_ks:.4f}, p-valor={p_ks:.4f}")
    print("   →", "Normal" if p_ks > 0.05 else "Não normal", "\n")
```

### 7.  Best Fit Distribution (usando diferentes métodos)

```python
import scipy.stats as st
import numpy as np
```

```python
# Função para encontrar a melhor distribuição
def best_fit_distribution(data, bins=200, ax=None):
    data = data.dropna()
    DISTRIBUTIONS = [
        st.norm, st.expon, st.gamma, st.beta,
        st.uniform, st.lognorm, st.weibull_min, st.weibull_max,
    ]
    best_distribution = None
    best_params = None
    best_sse = np.inf

    y, x = np.histogram(data, bins=bins, density=True)
    x_mid = (x + np.roll(x, -1))[:-1] / 2.0

    for distribution in DISTRIBUTIONS:
        try:
            params = distribution.fit(data)
            pdf = distribution.pdf(x_mid, *params[:-2], loc=params[-2], scale=params[-1])
            sse = np.sum(np.power(y - pdf, 2.0))

            if ax:
                ax.plot(x_mid, pdf, label=distribution.name)

            if sse < best_sse:
                best_distribution = distribution
                best_params = params
                best_sse = sse
        except Exception:
            continue

    return best_distribution.name, best_params

# Loop em todas as colunas numéricas do DataFrame
for coluna in df.select_dtypes(include=[np.number]).columns:
    print(f"\n Analisando coluna: {coluna}")
    dados = df[coluna].dropna()

    if len(dados) < 10:
        print("Coluna com poucos dados, pulando.")
        continue

    fig, ax = plt.subplots(figsize=(8, 5))
    ax.hist(dados, bins=50, density=True, alpha=0.5, label="Dados")

    melhor_dist, parametros = best_fit_distribution(dados, bins=50, ax=ax)

    ax.set_title(f"{coluna} - Melhor distribuição: {melhor_dist}")
    ax.set_xlabel(coluna)
    ax.set_ylabel("Densidade")
    ax.legend()
    ax.grid(True)
    plt.show()

    print(f"→ Melhor distribuição: {melhor_dist}")
    print(f"→ Parâmetros: {parametros}")
```

### 1.  O resultado do método V de Cramer

```python
import pandas as pd
import numpy as np
from scipy.stats import chi2_contingency
from tqdm import tqdm  # Para barra de progresso

# Função otimizada para V de Cramér
def cramers_v_optimized(x, y):
    confusion_matrix = pd.crosstab(x, y)
    if confusion_matrix.size == 0:
        return np.nan
    chi2 = chi2_contingency(confusion_matrix, correction=False)[0]
    n = confusion_matrix.sum().sum()
    phi2 = chi2 / n
    r, k = confusion_matrix.shape
    phi2_corr = max(0, phi2 - ((k - 1)*(r - 1)) / (n - 1))
    r_corr = r - ((r - 1)**2) / (n - 1)
    k_corr = k - ((k - 1)**2) / (n - 1)
    denom = min((k_corr - 1), (r_corr - 1))
    return np.sqrt(phi2_corr / denom) if denom > 0 else np.nan

# Parâmetros para controle
max_unique_categories = 280000

# Filtra colunas categóricas com limite de categorias únicas
categorical_cols = [
    col for col in df.select_dtypes(include=['object', 'category']).columns
    if df[col].nunique() <= max_unique_categories
]

# Calcular V de Cramér para pares únicos
results = []
for i, col1 in enumerate(tqdm(categorical_cols, desc="Calculando V de Cramér")):
    for col2 in categorical_cols[i+1:]:
        v = cramers_v_optimized(df[col1], df[col2])
        results.append((col1, col2, v))

# Converter para DataFrame
v_df = pd.DataFrame(results, columns=['Coluna 1', 'Coluna 2', 'V de Cramér'])
print(v_df.sort_values(by='V de Cramér', ascending=False))
```

```python
from sklearn.feature_extraction.text import CountVectorizer
import nltk
import string
from nltk.corpus import stopwords
```

```python
# Baixar stopwords (caso necessário)
nltk.download('stopwords')
stop_words = set(stopwords.words('portuguese'))

# 1. Filtrar mensagens de texto
df_texto = df[(df['has_media'] == 0) & (df['text_content_anonymous'].notnull())]
textos = df_texto['text_content_anonymous'].dropna().astype(str)

# 2. Pré-processamento: caixa baixa, remove pontuação, remove stopwords
def preprocessar(texto):
    texto = texto.lower()
    texto = texto.translate(str.maketrans('', '', string.punctuation))
    palavras = texto.split()
    palavras = [p for p in palavras if p not in stop_words and len(p) > 1]
    return ' '.join(palavras)

textos_limpos = textos.apply(preprocessar)

# 3. Função para extrair e plotar n-gramas
def top_ngrams(textos, n=1, top=30, titulo=''):
    vectorizer = CountVectorizer(ngram_range=(n, n))
    X = vectorizer.fit_transform(textos)
    sum_words = X.sum(axis=0)
    palavras = [(word, sum_words[0, idx]) for word, idx in vectorizer.vocabulary_.items()]
    palavras = sorted(palavras, key=lambda x: x[1], reverse=True)[:top]

    ngramas, contagens = zip(*palavras)

    plt.figure(figsize=(14, 8))
    ax = sns.barplot(x=list(contagens), y=list(ngramas), palette='mako')
    plt.title(f'Top {top} {titulo}', fontsize=16)
    plt.xlabel('Frequência', fontsize=14)
    plt.ylabel(f'{titulo}', fontsize=14)
    plt.grid(axis='x', linestyle='--', alpha=0.5)

    # Adicionando os valores na frente das barras
    for i, v in enumerate(list(contagens)):
        ax.text(v + max(contagens) * 0.01, i, str(v), color='black', va='center', fontsize=10)

    plt.tight_layout()
    plt.show()

# 4. Plotar unigramas, bigramas e trigramas
top_ngrams(textos_limpos, n=1, top=30, titulo='Unigramas')
top_ngrams(textos_limpos, n=2, top=30, titulo='Bigramas')
top_ngrams(textos_limpos, n=3, top=30, titulo='Trigramas')
```

### 32. A  rede  interativa  das  palavras  referente  às  mensagens  de  texto  (após  a remoção de stop words);

```python
from collections import Counter
import networkx as nx
from pyvis.network import Network
```

```python
# Baixar stopwords
nltk.download('stopwords')

# Filtrar só mensagens de texto
df_texto = df[(df['has_media'] == 0) & (df['text_content_anonymous'].notnull())]

# Preparar stopwords em português
stop_words = set(stopwords.words('portuguese'))

# Função para limpar e tokenizar texto
def preprocess(text):
    text = text.lower()
    text = re.sub(r'[^a-záéíóúãõçâêîôûàèìòù\s]', '', text)
    tokens = text.split()
    tokens = [t for t in tokens if t not in stop_words and len(t) > 1]
    return tokens

# Criar lista de tokens para cada mensagem
token_lists = df_texto['text_content_anonymous'].astype(str).map(preprocess)

# Construir contagem de co-ocorrências de palavras dentro da mesma mensagem
co_occur = Counter()
for tokens in token_lists:
    unique_tokens = set(tokens)
    for t1 in unique_tokens:
        for t2 in unique_tokens:
            if t1 != t2:
                pair = tuple(sorted([t1, t2]))
                co_occur[pair] += 1

# Filtrar as conexões mais fortes para não ficar poluído
limite = 5
co_occur_filtro = {k:v for k,v in co_occur.items() if v >= limite}

# Criar grafo
G = nx.Graph()
for (node1, node2), weight in co_occur_filtro.items():
    G.add_edge(node1, node2, weight=weight)

# Criar rede interativa com pyvis
net = Network(height='750px', width='100%', notebook=True, bgcolor='#222222', font_color='white')

# Adicionar nós e arestas ao pyvis
for node in G.nodes():
    net.add_node(node, label=node, title=node, size=15)

for source, target, data in G.edges(data=True):
    net.add_edge(source, target, value=data['weight'], title=f'Co-ocorrências: {data["weight"]}')

# Mostrar a rede (gera arquivo HTML e abre no navegador)
net.show('rede_palavras.html')
```

```python
# Garantir que a coluna date_message está no formato datetime
df['date_message'] = pd.to_datetime(df['date_message'], errors='coerce')

# Extrair o dia da semana (0=segunda, 6=domingo)
df['day_of_week'] = df['date_message'].dt.day_name()

# Contar quantidade de mensagens por dia da semana
mensagens_por_dia = df['day_of_week'].value_counts().reindex([
    'Monday', 'Tuesday', 'Wednesday', 'Thursday', 'Friday', 'Saturday', 'Sunday'
])

# Plotar gráfico de barras
plt.figure(figsize=(10, 6))
bars = plt.bar(mensagens_por_dia.index, mensagens_por_dia.values, color='cornflowerblue')
plt.title('Volume de Mensagens por Dia da Semana', fontsize=16)
plt.xlabel('Dia da Semana', fontsize=14)
plt.ylabel('Quantidade de Mensagens', fontsize=14)
plt.grid(axis='y', linestyle='--', alpha=0.5)

# Mostrar valores no topo das barras
for bar in bars:
    height = bar.get_height()
    plt.text(bar.get_x() + bar.get_width()/2, height + max(mensagens_por_dia.values)*0.01, 
             int(height), ha='center', va='bottom')

plt.tight_layout()
plt.show()
```

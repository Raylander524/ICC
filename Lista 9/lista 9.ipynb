```python
def monitor_memory(threshold_mb=10240, check_interval=5):
    proc = psutil.Process(os.getpid())
    while True:
        mem_mb = proc.memory_info().rss / (1024 ** 2)
        print(f"Memória atual: {mem_mb:.2f} MB")
        if mem_mb > threshold_mb:
            print(f"Memória excedeu {threshold_mb} MB. Encerrando o processo...")
            os._exit(1)
        time.sleep(check_interval)

# Execute essa função em uma thread separada
import threading
threading.Thread(target=monitor_memory, daemon=True).start()
```

```python
# Função para contar o número de palavras
def count_words(text):
    if isinstance(text, str):  # Verifica se o valor é uma string
        return len(text.split())
    return 0  # Caso não seja string, retorna 0 palavras

# Adiciona a coluna 'word_count' ao DataFrame
df['word_count'] = df['text_content_anonymous'].apply(count_words)

# Filtra os textos com mais de 5 palavras
df = df[df['word_count'] >= 5]
df
```

```python
def normalizar_texto(texto):
    texto = texto.lower()
    texto = re.sub(r"http\S+", "", texto)  # remove URLs
    texto = emoji.replace_emoji(texto, '')  # remove emojis
    texto = re.sub(r"[^a-zA-Z0-9\s]", "", texto)  # remove pontuação
    texto = unidecode.unidecode(texto)  # remove acentos
    texto = re.sub(r"\s+", " ", texto).strip()
    return texto

df['texto_normalizado'] = df['text_content_anonymous'].apply(normalizar_texto)

X_text_norm = df['texto_normalizado']  # coluna com os textos normalizados
```

```python
import numpy as np
from scipy.special import logsumexp

class GSDMM:
    def __init__(self, n_topics=10, alpha=0.1, beta=0.1, n_iter=30):
        self.n_topics = n_topics
        self.alpha = alpha
        self.beta = beta
        self.n_iter = n_iter
        self.eps = 1e-12  # Pequeno valor para estabilidade numérica

    def fit(self, bow_matrix, vocab):
        # Converter para matriz densa se esparsa
        if hasattr(bow_matrix, "toarray"):
            bow_matrix = bow_matrix.toarray()
        
        n_docs, n_words = bow_matrix.shape
        
        # Inicialização com verificação
        self.word_topic_counts = np.zeros((n_words, self.n_topics)) + self.beta
        self.topic_counts = np.zeros(self.n_topics) + self.alpha
        self.doc_topics = np.random.randint(0, self.n_topics, n_docs)
        
        # Gibbs Sampling com proteção numérica
        for iteration in range(self.n_iter):
            for i in range(n_docs):
                doc = bow_matrix[i]
                current_topic = self.doc_topics[i]
                
                # Remove documento do tópico atual (com proteção)
                self.word_topic_counts[:, current_topic] -= doc
                self.topic_counts[current_topic] -= 1
                
                # Garante contagens não-negativas
                np.maximum(self.word_topic_counts, self.eps, out=self.word_topic_counts)
                np.maximum(self.topic_counts, self.eps, out=self.topic_counts)
                
                # Cálculo estável das probabilidades
                log_probs = np.log(self.topic_counts + self.eps)
                log_probs += np.sum(
                    np.log(self.word_topic_counts.T + self.eps) * doc,
                    axis=1
                )
                
                # Normalização segura
                log_probs -= logsumexp(log_probs)
                probs = np.exp(log_probs)
                
                # Amostragem com fallback
                try:
                    new_topic = np.random.choice(self.n_topics, p=probs)
                except:
                    new_topic = np.argmax(probs)  # Fallback para o mais provável
                
                # Atualiza contagens
                self.word_topic_counts[:, new_topic] += doc
                self.topic_counts[new_topic] += 1
        
        self.vocab = vocab

    def get_topics(self, top_n=10):
        topics = {}
        for k in range(self.n_topics):
            top_words_idx = np.argsort(-self.word_topic_counts[:, k])[:top_n]
            topics[k] = [self.vocab[idx] for idx in top_words_idx]
        return topics

# Uso:
gsdmm = GSDMM(n_topics=5)
gsdmm.fit(bow_unigrama_norm_amostra, vocab_amostra)
topicos = gsdmm.get_topics()

# for topico, palavras in topicos.items():
#     print(f"Tópico {topico}: {', '.join(palavras)}")
```

```python
import pandas as pd
import matplotlib.pyplot as plt
import numpy as np

class GSDMM_Visualizer:
    def __init__(self, model, bow_matrix, texts, vocab):
        self.model = model
        self.bow_matrix = bow_matrix.toarray() if hasattr(bow_matrix, "toarray") else bow_matrix
        self.texts = texts  # Lista de textos originais
        self.vocab = vocab
        
    def get_topic_distribution(self):
        """Calcula a distribuição de tópicos para cada documento."""
        topic_dist = np.zeros((len(self.texts), self.model.n_topics))
        for i in range(len(self.texts)):
            doc = self.bow_matrix[i]
            log_probs = np.log(self.model.topic_counts + self.model.eps)
            log_probs += np.sum(
                np.log(self.model.word_topic_counts.T + self.model.eps) * doc,
                axis=1
            )
            probs = np.exp(log_probs - logsumexp(log_probs))
            topic_dist[i] = probs
        return topic_dist
    
    def visualize_topics(self, top_n_words=10):
        """Mostra palavras-chave por tópico e distribuição global."""
        # Palavras-chave por tópico
        topics = self.model.get_topics(top_n_words)
        print("\n🔍 Palavras-chave por Tópico:")
        for topic, words in topics.items():
            print(f"Tópico {topic}: {', '.join(words)}")
        
        # Distribuição de documentos por tópico
        topic_dist = self.get_topic_distribution()
        dominant_topics = np.argmax(topic_dist, axis=1)
        topic_counts = pd.Series(dominant_topics).value_counts().sort_index()
        
        # plt.figure(figsize=(10, 5))
        # topic_counts.plot(kind='bar', color='skyblue')
        # plt.title("📊 Distribuição de Documentos por Tópico")
        # plt.xlabel("Tópico")
        # plt.ylabel("Número de Documentos")
        # plt.xticks(rotation=0)
        # plt.show()
        
    def visualize_document_topics(self, n_docs=5):
        """Mostra exemplos de documentos e seus tópicos principais."""
        topic_dist = self.get_topic_distribution()
        dominant_topics = np.argmax(topic_dist, axis=1)
        top_percentages = np.max(topic_dist, axis=1)
        
        # Criar DataFrame com resultados
        df = pd.DataFrame({
            "Texto": self.texts,
            "Tópico_Principal": dominant_topics,
            "Porcentagem": top_percentages.round(3)
        })
        
        print("\n📄 Exemplos de Documentos Classificados:")
        print(df.head(n_docs))
        
        # Gráfico de distribuição para um documento específico
        doc_id = 0  # Altere para ver outros documentos
        # plt.figure(figsize=(8, 4))
        # plt.bar(range(self.model.n_topics), topic_dist[doc_id], color='orange')
        # plt.title(f"Distribuição de Tópicos para Documento #{doc_id}")
        # plt.xlabel("Tópico")
        # plt.ylabel("Probabilidade")
        # plt.show()

# Uso:
visualizer = GSDMM_Visualizer(gsdmm, bow_unigrama_norm_amostra, X_text_norm, vocab_amostra)
visualizer.visualize_topics()
visualizer.visualize_document_topics(n_docs=10)
```
